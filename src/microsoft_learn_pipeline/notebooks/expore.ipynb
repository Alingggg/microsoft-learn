```python
import json
from pathlib import Path

catalog_filename = "microsoft_learn_catalog_20260918.json"
catalog_path = next(
    (directory / catalog_filename for directory in (Path.cwd(), *Path.cwd().parents)
     if (directory / catalog_filename).is_file()),
    None,
)
if catalog_path is None:
    raise FileNotFoundError(f"Could not find {catalog_filename} above {Path.cwd()}")

with catalog_path.open(encoding="utf-8") as file:
    data = json.load(file)

print(f"Loaded {catalog_path}")
print(f"Top-level sections: {', '.join(data)}")
print(f"Modules: {len(data.get('modules', [])):,}")
```

```python
top_level_sections = ["units", "learningPaths", "appliedSkills", "certifications", "mergedCertifications", "exams", "courses", "levels", "products", "roles", "subjects"]

for section in top_level_sections:
    section_sample = data[section][0]
    print(f"======== {section} ========")

    for key, value in section_sample.items():
        print(f"{key}: {type(value)}")

    print()
```
